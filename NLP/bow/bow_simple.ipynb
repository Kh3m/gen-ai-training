### Simple Example

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction import text
import pandas as pd
```

```python
print(sorted(text.ENGLISH_STOP_WORDS))
```

```python
sentences = [
    "The food is good",
    "The food is not good good",
]

sentences
```

```python
vectorizer = CountVectorizer(
    stop_words="english"
)

X = vectorizer.fit_transform(
    sentences
)

X 
```

```python
print(vectorizer.get_feature_names_out())
print(vectorizer.vocabulary_)
print(X.toarray())
```

```python
X.toarray()
```

```python
df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=sentences,
)

cleaned  = []

lower_sentences = [ sent.lower() for sent in sentences ]

for sent in lower_sentences:
    words = sent.split(" ")
    sent = [ word for word in words if word not in text.ENGLISH_STOP_WORDS ]  
    cleaned.append(" ".join(sent))
        
df.insert(0, "cleaned", cleaned)
df
```
