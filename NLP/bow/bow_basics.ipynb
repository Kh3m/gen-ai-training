### Simple Example

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction import text
import pandas as pd
```

```python
print(sorted(text.ENGLISH_STOP_WORDS))
```

```python
sentences = [
    "The food is good",
    "The food is not good good",
]

sentences
```

```python
vectorizer = CountVectorizer(
    stop_words="english"
)

X = vectorizer.fit_transform(
    sentences
)

X 
```

```python
print(vectorizer.get_feature_names_out())
print(vectorizer.vocabulary_)
print(X.toarray())
```

```python
X.toarray()
```

```python
df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=sentences,
)

def clean_data(sentences):
    
    cleaned  = []

    lower_sentences = [ sent.lower() for sent in sentences ]

    for sent in lower_sentences:
        words = sent.split(" ")
        sent = [ word for word in words if word not in text.ENGLISH_STOP_WORDS ]  
        cleaned.append(" ".join(sent))
        
    return cleaned
        
df.insert(0, "cleaned", clean_data(sentences))
df
```

### Example 2 with n_grams

```python
sentences_2 = [
    "He is a good Boy",
    "She is a good girl",
    "Boy and girl are good",
    "I love boy and girl",
    "just extras to increase sentences to get to four n_grams"
]
sentences_2
```

```python
vectorizer2 = CountVectorizer(
    stop_words="english",
)
vectors = vectorizer2.fit_transform(sentences_2)
vectors
```

```python
feature_names2 = vectorizer2.get_feature_names_out()
feature_names2
```

```python
vectors.toarray()
```

```python
df_2 = pd.DataFrame(
    data=vectors.toarray(),
    columns=feature_names2,
    index=sentences_2
)

df_2.insert(0, "cleaned data", clean_data(sentences_2))
df_2
```

```python
def with_n_grams(n_grams=(1,1)):
    vectorizer = CountVectorizer(
        stop_words="english",
        ngram_range=n_grams,
    )
    vectors = vectorizer.fit_transform(sentences_2)
    feature_names = vectorizer.get_feature_names_out()
    
    df_2 = pd.DataFrame(
        data=vectors.toarray(),
        columns=feature_names,
        index=clean_data(sentences_2)
    )

    return df_2
```

```python
# (1,1) is the default: every feature is a single word — a unigram
with_n_grams()
```

```python
# (1,2): unigrams + bigrams — single words and adjacent word pairs
with_n_grams(n_grams=(1,2))
```

```python
# (1,3): unigrams + bigrams + trigrams — runs of 1, 2, and 3 adjacent words
with_n_grams(n_grams=(1,3))
```

```python
# (2,3): bigrams + trigrams only — single words are dropped
with_n_grams(n_grams=(2,3))
```

```python
# (2,4): bigrams, trigrams, and 4-grams
with_n_grams(n_grams=(2,4))
```
