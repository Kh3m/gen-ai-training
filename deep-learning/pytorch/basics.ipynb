### Tensor operations

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])

print(a + b)          # element-wise add
print(a * b)          # element-wise multiply
print(a @ b)          # dot product (matrix mult for 1D)
```

```python
x = torch.rand(2, 3)
print(x.sum())        # sum all elements
print(x.mean())       # mean
print(x.view(3, 2))   # reshape (like numpy's reshape)
print(x.T)            # transpose
```
