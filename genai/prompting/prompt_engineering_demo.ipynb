```python
client = Groq(api_key=api_key)

if api_key is None:
  raise ValueError('GROQ_API_KEY not found in Colab Secrets')

print('API key loaded successfully')
```

```python
def run_prompt(prompt, temperature=1, max_tokens=200):
  response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    temperature=temperature,
    max_completion_tokens=max_tokens,
    messages=[
        {
            "role": "system",
            "content": "You're an expert output summarizer from generated LLM response. If the response is above 200 words, summarize to below 200"
        },
        {
            "role": "user",
            "content": prompt
        }
    ]
    )
  return response.choices[0].message.content
```

```python
prompt_v1 = 'Explain AI.'
run_prompt(prompt_v1)
```

```python
prompt_v2 = (
'You are a teacher. '
'Explain artificial intelligence to a beginner.'
)

run_prompt(prompt_v2)
```

```python
prompt_v3 = (
'You are a teacher explaining concepts to beginners. '
'Explain artificial intelligence in simple language. '
'Limit the explanation to 100 words.'
)
print(run_prompt(prompt_v3))
```

```python
prompt_v4 = (
'You are a teacher explaining concepts to beginners. '
'Explain artificial intelligence in simple language for a non-technical audience. '
'Use one real-world example. '
'Limit the explanation to 100 words.'
)
print(run_prompt(prompt_v4))
```

```python
prompt_v5 = (
'You are a teacher explaining concepts to beginners. '
'Explain artificial intelligence in simple language for a non-technical audience. '
'Use exactly one real-world example. '
'Structure the answer in two short paragraphs. '
'Limit the explanation to 100 words.'
)
print(run_prompt(prompt_v5))
```

```python
prompts = {
    'V1 - Poor': prompt_v1,
    'V2 - Role': prompt_v2,
    'V3 - Constraints': prompt_v3,
    'V4 - Context': prompt_v4,
    'V5 - Structure': prompt_v5,
}

for k, p in prompts.items():
  print(f'\n{k}\n{'-'*50}')
  print(run_prompt(p))
```
